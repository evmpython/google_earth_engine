# Carregando os dados do ERA5 mensal e transformando para dataset do xarray com a biblioteca Xee
 - https://developers.google.com/earth-engine/datasets/catalog/ECMWF_ERA5_LAND_MONTHLY_AGGR?hl=pt-br

```python
%%time
# carrega dados ERA5 mensais
era5 = ee.ImageCollection('ECMWF/ERA5_LAND/MONTHLY_AGGR').filterDate('2000-01-01','2025-09-01')

# converter para Dataset do Xarray
ds = xr.open_dataset(era5, engine='ee', crs='EPSG:4326', scale=0.25)

# mostrando os dados
ds
```

```python
# mapa interativo com geemap
Map = geemap.Map(basemap='SATELLITE')
Map
```

```python
# selecionando região de forma interativa no Mapa. Clicar no mapa e depois rodar essa célula de código
roi = Map.draw_last_feature.geometry()
roi
```

```python
%%time

# região de estudo
#roi = ee.FeatureCollection('FAO/GAUL/2015/level0').filter(ee.Filter.eq('ADM0_NAME', 'Brazil')) # Em "geometry" usar geometry=roi.geometry()
#roi = ee.FeatureCollection('FAO/GAUL/2015/level1').filter(ee.Filter.eq('ADM1_NAME', 'Sao Paulo')) # Em "geometry" usar geometry=roi.geometry()
#roi = ee.FeatureCollection('FAO/GAUL/2015/level2').filter(ee.Filter.eq('ADM2_NAME', 'Sao Paulo')) # Em "geometry" usar geometry=roi.geometry()
#roi = ee.Geometry.Rectangle([-47.1, -21.1, -44.2, -20.4]) # Em "geometry" usar geometry=roi

# carrega os dados de t2m
t2m_mensal = ee.ImageCollection('ECMWF/ERA5_LAND/MONTHLY_AGGR')\
               .select('temperature_2m')\
               .filterDate('2010-01-01','2024-12-31')\
               .map(lambda img: img.subtract(273.15).copyProperties(img, img.propertyNames()))

# converte para Dataset
ds_t2m_mensal = xr.open_dataset(t2m_mensal,
                                engine = 'ee',
                                crs = 'EPSG:4326',
                                scale = 0.11,
                                geometry = roi)

# muda de "(time, lon, lat)" para "(time, lat, lon)"
#ds_t2m_mensal = ds_t2m_mensal.transpose("time", "lat", "lon")

# mostrando os dados
ds_t2m_mensal
```

```python
# mostra os valores mínimos e máximos de temperatura do dataset mensal
print(ds_t2m_mensal['temperature_2m'].min().values, ds_t2m_mensal['temperature_2m'].max().values)
```

```python
# média anual de temperatura
ds_t2m_anual = ds_t2m_mensal.groupby('time.year').mean('time')
ds_t2m_anual
```

```python
# plota mapa de temperatura média anual
ds_t2m_anual['temperature_2m'].plot(x = 'lon',
                                    y = 'lat',
                                    col = 'year',
                                    robust = True,
                                    col_wrap = 5,
                                    cmap = 'coolwarm')

# salva figura
plt.savefig('temp_anual.png', dpi=300, bbox_inches='tight')
```

```python
# calcula a anomalia anual
ds_t2m_anual_anomalia = ds_t2m_anual - ds_t2m_anual.mean('year')
ds_t2m_anual_anomalia
```

```python
# plota mapa de anomalia média anual
ds_t2m_anual_anomalia['temperature_2m'].plot(x = 'lon',
                                             y = 'lat',
                                             col = 'year',
                                             robust = True,
                                             col_wrap = 5,
                                             cmap = 'coolwarm')

# salva figura
plt.savefig('anom_anual.png', dpi=300, bbox_inches='tight')
```

```python
# série temporal de temperatura para um ponto (lat, lon)
ds_t2m_anual_sub = ds_t2m_anual.sel(lat=-15.5, lon=-60.2, method='nearest')
ds_t2m_anual_sub
```

```python
# plota série temporal
ds_t2m_anual_sub['temperature_2m'].plot()

# salva figura
plt.savefig('ts_anual.png', dpi=300, bbox_inches='tight')
```

```python
# salva arquivos em formato NetCDF
ds_t2m_mensal.to_netcdf('t2m_mensal.nc')
ds_t2m_anual.to_netcdf('t2m_anual.nc')
ds_t2m_anual_anomalia.to_netcdf('anom_anual.nc')
```

# Plotando mapa com o ultraplot

```python
# instalando e importando o proplot
!pip install -q ultraplot cartopy
import ultraplot as uplt
```

```python
%%time
# região de estudo: escolhemos o Estado de SP
estado_SP = ee.FeatureCollection('FAO/GAUL/2015/level1').filter(ee.Filter.eq('ADM1_NAME', 'Sao Paulo')) # Em "geometry" usar geometry=roi.geometry()

# carrega os dados de t2m
t2m_mensal = ee.ImageCollection('ECMWF/ERA5_LAND/MONTHLY_AGGR') \
               .filterDate('2010-01-01','2024-12-31') \
               .filterBounds(estado_SP) \
               .select('temperature_2m') \
               .map(lambda img: img.subtract(273.15).copyProperties(img, img.propertyNames()))

# converte para Dataset
ds_t2m_mensal = xr.open_dataset(t2m_mensal,
                                engine = 'ee',
                                crs = 'EPSG:4326',
                                scale = 0.11,
                                geometry = estado_SP.geometry())

# muda de "(time, lon, lat)" para "(time, lat, lon)"
ds_t2m_mensal_transposto = ds_t2m_mensal.transpose("time", "lat", "lon")
```

```python
# mostrando os dados originais no formato (time, lon, lat)
ds_t2m_mensal
```

```python
# mostra os dados transposto no formato (time, lat, lon)
ds_t2m_mensal_transposto
```

```python
%%time
# média anual de temperatura
ds_t2m_anual_transposto = ds_t2m_mensal_transposto.groupby('time.year').mean('time')
ds_t2m_anual_transposto
```

```python
t2m_2024_sp = ds_t2m_anual_transposto.sel(year=2024)
t2m_2024_sp
```

```python
# importa bibliotecas
import cartopy.io.shapereader as shpreader
import cartopy.crs as ccrs

# extraindo o ano de 2024
t2m_2024_sp = ds_t2m_anual_transposto.sel(year=2024)

# moldura da figura
fig, ax = uplt.subplots(axheight=5, tight=True, proj='pcarree')

# limites dos dados
latmin, latmax = float(t2m_2024_sp['lat'][0]), float(t2m_2024_sp['lat'][-1])
lonmin, lonmax = float(t2m_2024_sp['lon'][0]), float(t2m_2024_sp['lon'][-1])

# mapa de contornos preenchidos
map1 = ax.contourf(t2m_2024_sp['lon'],
                   t2m_2024_sp['lat'],
                   t2m_2024_sp['temperature_2m'],
                   cmap='coolwarm',
                   levels=uplt.arange(10, 40, 5))

# formatação dos eixos
ax.format(coast=False, borders=False, innerborders=False, labels=True,
          latlines=2, lonlines=2,
          lonlim=(lonmin-0.09, lonmax+0.09),
          latlim=(latmin-0.09, latmax+0.09),
          title='Temperatura em 2m do ERA5 para o Estado de SP: 2024',
          small='20px', large='20px')

# plota contorno do RS
shapefile = list(shpreader.Reader('https://github.com/evmpython/shapefile/raw/refs/heads/main/UFs/SP/SP_UF_2019.shp').geometries())
ax.add_geometries(shapefile, ccrs.PlateCarree(), edgecolor='black', facecolor='none', linewidth=1.0, alpha=1.0)

# barra de cores
fig.colorbar(map1, loc='r', label='Temperatura [${^O}$ C]', ticks=5, ticklabelsize=13, labelsize=13, space=0.1)

# salvar a figura
fig.save('mapa_ERA5_SP.jpeg', dpi=300)
```
