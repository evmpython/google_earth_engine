#**Exportando FeatureCollection para SHP, GEOSON ou CSV**


# **1° Passo:** Inicializando GEE

```python
# importando o GEE e Geemap
import ee
import geemap

# inicializando GEE
geemap.ee_initialize(project='ee-enrique')
```

#**Exportando uma `ee.FeatureCollection`**


```python
# define região
estado_mg = ee.FeatureCollection('FAO/GAUL/2015/level1').filter(ee.Filter.eq('ADM1_NAME', 'Minas Gerais'))

# criando um mapa interativo
Map = geemap.Map()

# centra o mapa na região
Map.centerObject(estado_mg, zoom=6)

# adicionando basemap
Map.add_basemap('SATELLITE')

# contorno do Estado de MG
style = {'color': 'yellow', 'fillColor': '00000000'}
Map.addLayer(estado_mg.style(**style), {}, 'MG')

# exibe na tela
Map
```

```python
# formato SHP
geemap.ee_to_shp(estado_mg, filename="MG.shp")
```

```python
# formato GEOSON
geemap.ee_to_geojson(estado_mg, filename="MG.geojson")
```

```python
# formato CSV
geemap.ee_to_csv(estado_mg, filename="MG.csv")
```
