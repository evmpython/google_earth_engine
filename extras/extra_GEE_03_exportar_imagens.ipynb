# **1° Passo:** Inicializando GEE

```python
# importando o GEE e Geemap
import ee
import geemap

# inicializando GEE
geemap.ee_initialize(project='ee-enrique')
```

#**Exportando uma `ee.Image`**

```python
Map = geemap.Map()
image = ee.Image("LANDSAT/LC08/C02/T1_TOA/LC08_044034_20140318").select(["B5", "B4", "B3"])
vis_params = {"min": 0, "max": 0.5, "gamma": [0.95, 1.1, 1]}
Map.center_object(image, 8)
Map.add_layer(image, vis_params, "Landsat")
Map
```

Adicion um retangulo ao map

```python
region = ee.Geometry.BBox(-122.5955, 37.5339, -122.0982, 37.8252)
fc = ee.FeatureCollection(region)
style = {"color": "ffff00ff", "fillColor": "00000000"}
Map.add_layer(fc.style(**style), {}, "ROI")
```

Salva

```python
geemap.ee_export_image(image, filename="landsat.tif", scale=30, region=region)
```

Verificando a projeção

```python
projection = image.select(0).projection().getInfo()
projection
```

```python
crs = projection["crs"]
crs_transform = projection["transform"]
```

Região específica, crs, e crs_transform

```python
geemap.ee_export_image(image,
                       filename="landsat_crs.tif",
                       crs=crs,
                       crs_transform=crs_transform,
                       region=region)
```

Salva no Drive

```python
geemap.ee_export_image_to_drive(image, description="landsat", folder="export", region=region, scale=30)
```

#**Exportando uma `ee.ImageCollection`**

```python
point = ee.Geometry.Point(-99.2222, 46.7816)
collection = ee.ImageCollection("USDA/NAIP/DOQQ") \
               .filterBounds(point) \
               .filterDate("2008-01-01", "2018-01-01") \
               .filter(ee.Filter.listContains("system:band_names", "N"))
```

```python
collection.aggregate_array("system:index")
```

```python
geemap.ee_export_image_collection(collection, out_dir=".", scale=10)
```
